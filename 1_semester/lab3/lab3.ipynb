# Импорты

```python
import re
from nltk.stem.snowball import SnowballStemmer
from collections import Counter
```

# Загрузка текста

```python
with open('Platon-Menon.txt', 'r', encoding='windows-1251') as f:
    text = f.read()

text
```

# Обработка текста

## 1. Приведение к нижнему регистру

```python
text = text.lower()
text
```

## 2. Удаление знаков препинания

```python
text = re.sub(r"[^а-яa-z\s]", "", text)
text
```

## 3. Удаление лишних пробелов

```python
text = re.sub(r'\s+', ' ', text).strip()
text
```

## 4. Удаление мусора

### Удаление предисловия

```python
preface = "спасибо что скачали книгу в бесплатной электронной библиотеке royallibru httproyallibru все книги автора httproyallibruauthorplatonhtml эта же книга в других форматах httproyallibrubookplatonmenonhtml приятного чтения платон менон платон менон менон сократ раб менона анит "
text = text.replace(preface, "")
text
```

### Удаление послесловия

```python
afterword = " спасибо что скачали книгу в бесплатной электронной библиотеке royallibru httproyallibru оставить отзыв о книге httproyallibrucommentplatonmenonhtml все книги автора httproyallibruauthorplatonhtml"
text = text.replace(afterword, "")
text
```

## 5. Разбиение слов

```python
words = text.split()
words
```

## 6. Стэмминг

```python
stemmer = SnowballStemmer("russian")
stems = [stemmer.stem(w) for w in words]
stems
```

## 7. Токенизация (мешок слов)

```python
bag_of_words = Counter(stems)
bag_of_words
```
