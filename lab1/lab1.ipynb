```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score
from sklearn.ensemble import RandomForestClassifier
```

# Загрузка данных

```python
df = pd.read_csv("student_depression_dataset.csv", index_col=0)
df
```

```python
Y = df['Depression']
df.drop(['Depression'],inplace=True,axis=1)
df.head()
```

# Удаление лишних столбцов

```python
df.drop(['City'],axis=1,inplace=True)
df.head()
```

# Обработка пропусков

```python
df.isna().sum()
```

```python
Y.isna().sum()
```

# Перевод категориальных признаков в численные

```python
for i in df.columns:
    print(i)
    print(df[i].unique())
```

## Label encode

```python
def convert_sleep_duration_to_int(x):
    if x == "'Less than 5 hours'":
        return 1
    elif x == "'5-6 hours'":
        return 2
    elif x == "'7-8 hours'":
        return 3
    elif x == "'More than 8 hours'":
        return 4
    
    return np.nan

df["Sleep Duration"] = df["Sleep Duration"].apply(convert_sleep_duration_to_int).astype("Int64")
df.head()
```

```python
def convert_dietary_habits_to_int(x):
    if x == 'Unhealthy':
        return 1
    elif x == 'Moderate':
        return 2
    elif x == 'Healthy':
        return 3
    
    return np.nan

df["Dietary Habits"] = df["Dietary Habits"].apply(convert_dietary_habits_to_int).astype("Int64")
df.head()
```

```python
df["Have you ever had suicidal thoughts ?"] = df["Have you ever had suicidal thoughts ?"].apply(lambda x: 1 if x == 'Yes' else 0).astype(int)
df.head()
```

```python
df["Family History of Mental Illness"] = df["Family History of Mental Illness"].apply(lambda x: 1 if x == 'Yes' else 0).astype(int)
df.head()
```

## One hot encode

```python
df = pd.get_dummies(df, columns=["Gender", "Profession", "Degree"], dtype=int, drop_first=True)
df.info()
```

## Обработка пропусков

```python
df.isna().sum()
```

```python
df["Sleep Duration"] = df["Sleep Duration"].fillna(df["Sleep Duration"].median())
df["Dietary Habits"] = df["Dietary Habits"].fillna(df["Dietary Habits"].median())

df.isna().sum()
```

# Обработка числовых признаков

## Объединение схожих столбцов

```python
df[df["Work Pressure"] != 0]
```

```python
df["Pressure"] = df["Academic Pressure"] + df["Work Pressure"]
df.drop(["Academic Pressure", "Work Pressure"], inplace=True, axis=1)
df.head()
```

```python
df[df["Job Satisfaction"] != 0]
```

## Обработка выбросов

```python
df["Financial Stress"] = df["Financial Stress"].apply(lambda x: np.nan if x == "?" else x).astype("Float64")
df.head()
```

```python
df.isna().sum()
```

```python
df["Financial Stress"] = df["Financial Stress"].fillna(df["Financial Stress"].median())
df.isna().sum()
```

# Нормализация

```python
df_norm = (df - df.min()) / (df.max() - df.min())
df_norm
```

# Обучение модели

```python
X_train, X_test, Y_train, Y_test = train_test_split(df_norm, Y, test_size=0.2, random_state=42, stratify=Y)
```

```python
model = RandomForestClassifier(max_depth=10, random_state=42)
model.fit(X_train, Y_train)
```

```python
Y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(Y_test, Y_pred)
print("Accuracy:", accuracy)

precision = precision_score(Y_test, Y_pred)
print("Precision:", precision)
```
